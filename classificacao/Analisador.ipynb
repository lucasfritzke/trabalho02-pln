```python
import pandas as pd
import numpy as np

# Lê o arquivo CSV
df = pd.read_csv('comentarios05.csv')

# Define as condições e os novos valores
condicoes = [
    df['nota_comentario'] <= 3,
    (df['nota_comentario'] > 4.0) & (df['nota_comentario'] <= 5.0)
]
valores = [1, 2,3]

# Aplica a transformação
df['nota_comentario'] = np.select(condicoes, valores)

# Exibe a contagem de valores
print(df['nota_comentario'].value_counts())

# Salva as alterações no mesmo arquivo (sem o índice)
df.to_csv('comentarios02.csv', index=False)

print("Alterações salvas com sucesso em 'comentarios02.csv'")
```

```python
# python
import pandas as pd
import numpy as np

# lê CSV
df = pd.read_csv('comentarios02.csv')

# garante que a nota seja numérica (troca vírgula por ponto) e marca erros como NaN
df['nota_comentario'] = pd.to_numeric(
    df['nota_comentario'].astype(str).str.replace(",", ".", regex=False),
    errors='coerce'
)

# remove linhas sem nota
df = df.dropna(subset=['nota_comentario'])

# mapear em duas classes: 1 = <=4.0 (negativo/baixo), 2 = >=4.5 (positivo/alto)
df['nota_comentario'] = np.where(df['nota_comentario'] >= 4.5, 2, 1).astype(int)

# exibe contagem e salva
print(df['nota_comentario'].value_counts())

df.to_csv('comentarios02.csv', index=False)
print("Alterações salvas com sucesso em `comentarios02.csv`")
```

```python
# 5 classes: 1 (<=1.5), 2 (<=2.5), 3 (<=3.5), 4 (<=4.5), 5 (>4.5)
import pandas as pd
import numpy as np

df = pd.read_csv('comentarios05.csv')
df['nota_comentario'] = pd.to_numeric(df['nota_comentario'].astype(str).str.replace(",", ".", regex=False), errors='coerce')
df = df.dropna(subset=['nota_comentario'])

bins_5 = [0, 1.5, 2.5, 3.5, 4.5, np.inf]
labels_5 = [1, 2, 3, 4, 5]
df['nota_comentario'] = pd.cut(df['nota_comentario'], bins=bins_5, labels=labels_5).astype(int)

print(df['nota_comentario'].value_counts())
df.to_csv('comentarios05.csv', index=False)
print("Alterações salvas com sucesso em `comentarios05.csv` (5 classes)")
```

```python
# 3 classes: 1 (<=3.0), 2 (3.1 a 4.4), 3 (>=4.5)
import pandas as pd
import numpy as np

df = pd.read_csv('comentarios03.csv')
df['nota_comentario'] = pd.to_numeric(df['nota_comentario'].astype(str).str.replace(",", ".", regex=False), errors='coerce')
df = df.dropna(subset=['nota_comentario'])

# python
condicoes = [
    df['nota_comentario'] <= 3.5,
    (df['nota_comentario'] > 3.5) & (df['nota_comentario'] <= 4.5),
    df['nota_comentario'] > 4.5
]
valores = [1, 2, 3]

df['nota_comentario'] = np.select(condicoes, valores)

print(df['nota_comentario'].value_counts())
df.to_csv('comentarios03.csv', index=False)
print("Alterações salvas com sucesso em `comentarios02.csv` (3 classes)")
```
