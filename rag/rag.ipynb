```python
import pandas as pd
```

# Criando arquivo txt

```python
filmes = pd.read_csv('../filmes_processados_csv/filmes_processados.csv', usecols=['nome', 'resumo'])

with open('filmes.txt', 'w', encoding="utf-8") as arquivo:
    for filme in filmes.iterrows():
        resumo = filme[1]['resumo']
        nome = filme[1]['nome']
        arquivo.write(f"###{nome}\n{resumo}\n")
```

```python
from langchain_text_splitters.character import RecursiveCharacterTextSplitter

# Separando o texto em chunks de 400 caracteres ou quando encontra  `###`
with open('filmes.txt', 'r', encoding="utf-8") as arquivo:
    texto = arquivo.read()

text_splitter = RecursiveCharacterTextSplitter(
    separators=["###", " ", ""],
    chunk_size=400,
    chunk_overlap=0,
    length_function=len
)
docs = text_splitter.create_documents([texto])
```
