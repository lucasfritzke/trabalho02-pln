```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
# URL base do site AdoroCinema
URL = "https://www.adorocinema.com/filmes/melhores/adorocinema/?page="

# Lista para armazenar os filmes
MOVIES_DETAILS = []

# Loop para percorrer as primeiras 4 páginas
for i in range(1, 11):
    # Requisição HTTP para a página atual
    response = requests.get(URL + str(i))
    content = response.content

    soup = BeautifulSoup(content, 'html.parser')

    links = soup.find_all('a', class_='meta-title-link')

    for link in links:
        url = 'https://www.adorocinema.com' + link.get('href')
        response = requests.get(url + 'criticas-adorocinema/')
        content = response.content

        soup = BeautifulSoup(content, 'html.parser')

        details = soup.find_all('p', class_='bo-p')
        resumo = ''
        for detail in details:
            resumo += detail.getText()

        title = soup.find('div', class_='title').getText()

        MOVIES_DETAILS.append({
            'nome': title.strip(),
            'resumo': resumo.strip()
        })



    print(f"Página {i} processada. Total de filmes até agora: {len(MOVIES_DETAILS)}")
```

# Criando arquivo txt

```python
for i in range(len(MOVIES_DETAILS)):
    filme = MOVIES_DETAILS[i]
    with open(f'filmestxts/{MOVIES_DETAILS[i]['nome']}.txt', 'w', encoding="utf-8") as arquivo:
        resumo = filme['resumo']
        nome = filme['nome']
        arquivo.write(f"{nome}\n{resumo}\n")
```

```python
from langchain_text_splitters.character import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import TextLoader
import os

docs = []

nomes_arquivos = os.listdir('filmestxts/')
print(nomes_arquivos)
for nome_arquivo in nomes_arquivos:
    loader = TextLoader(f"filmestxts/{nome_arquivo}", encoding="utf-8")
    documentos = loader.load()

    text_splitter = RecursiveCharacterTextSplitter(
        separators=[" ", ""],
        chunk_size=500,
        chunk_overlap=50,
        length_function=len,
    )
    docs.append(text_splitter.split_documents(documentos))

docs
```
